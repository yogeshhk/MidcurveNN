# EquiCollapse: extended comparison

Reads `runs/extended_metrics.json` (written by `evaluate_extended.py`) and prints one table. Trains nothing. See `../../../proposal/RESULTS_gdl.md` for caveats.

```python
import json, os
path = os.path.join('runs', 'extended_metrics.json')
data = json.load(open(path))
models = list(data)
tests = []
for m in models:
    for t in data[m]:
        if t not in tests:
            tests.append(t)
print('%-16s' % 'test' + ''.join('%-40s' % m[:38] for m in models))
for t in tests:
    row = '%-16s' % t
    for m in models:
        v = data[m].get(t)
        row += '%-40s' % ('%.3f' % v['ok_025w'] if v else '-')
    print(row)
```
